## **Removing variables that could cause data leakage**

```r
hotel$children[is.na(hotel$children)] <- 0

hotel_model <- hotel[, c(
  "is_canceled",
  "hotel",
  "lead_time",
  "arrival_date_year",
  "arrival_date_month",
  "stays_in_weekend_nights",
  "stays_in_week_nights",
  "adults",
  "children",
  "babies",
  "meal",
  "market_segment",
  "distribution_channel",
  "is_repeated_guest",
  "previous_cancellations",
  "previous_bookings_not_canceled",
  "reserved_room_type",
  "assigned_room_type",
  "booking_changes",
  "deposit_type",
  "days_in_waiting_list",
  "customer_type",
  "adr",
  "required_car_parking_spaces",
  "total_of_special_requests"
)]

hotel_model$is_canceled <- factor(
  hotel_model$is_canceled,
  levels = c(0, 1),
  labels = c("Not_Canceled", "Canceled")
)

str(hotel_model)
summary(hotel_model)
```

# **Logistic Regression to predict whether a booking will be canceled**

```r
install.packages(c("caret", "pROC"))
library(caret)
library(pROC)
```

# **Split the data into training (80%) and testing (20%)**

```r
library(caret)
library(pROC)
```

```r
set.seed(123)

train_index <- createDataPartition(
  hotel_model$is_canceled,
  p = 0.80,
  list = FALSE
)

train_data <- hotel_model[train_index, ]
test_data <- hotel_model[-train_index, ]

dim(train_data)
dim(test_data)

prop.table(table(train_data$is_canceled))
prop.table(table(test_data$is_canceled))
```

# **Logistic Regression with 5-fold Cross-Validation**

```r
set.seed(123)

control <- trainControl(
  method = "cv",
  number = 5,
  classProbs = TRUE,
  summaryFunction = twoClassSummary,
  savePredictions = "final"
)

logistic_model <- train(
  is_canceled ~ .,
  data = train_data,
  method = "glm",
  family = binomial,
  metric = "ROC",
  trControl = control
)

logistic_model
```

# **20% test dataset**

```r
pred_prob <- predict(
  logistic_model,
  newdata = test_data,
  type = "prob"
)

pred_class <- predict(
  logistic_model,
  newdata = test_data,
  type = "raw"
)

confusionMatrix(
  pred_class,
  test_data$is_canceled,
  positive = "Canceled"
)
```

# **ROC curve and AUC on the test data**

```r
roc_curve <- roc(
  test_data$is_canceled,
  pred_prob$Canceled,
  levels = c("Not_Canceled", "Canceled"),
  direction = "<"
)

auc_value <- auc(roc_curve)

auc_value

plot(
  roc_curve,
  main = "ROC Curve - Logistic Regression"
)
```

```r
optimized_data <- train_data[, c(
  "is_canceled",
  "hotel",
  "lead_time",
  "arrival_date_year",
  "arrival_date_month",
  "stays_in_weekend_nights",
  "stays_in_week_nights",
  "adults",
  "children",
  "is_repeated_guest",
  "previous_cancellations",
  "previous_bookings_not_canceled",
  "booking_changes",
  "deposit_type",
  "customer_type",
  "adr",
  "total_of_special_requests"
)]

optimized_data <- droplevels(optimized_data)

set.seed(123)

optimized_model <- train(
  is_canceled ~ .,
  data = optimized_data,
  method = "glm",
  family = binomial,
  metric = "ROC",
  trControl = control
)

optimized_model
```
