Outlier Detection

```r
boxplot(hotel_data$adr,
        main = "Boxplot of Average Daily Rate (ADR)",
        ylab = "ADR")
```

Detect Outliers Using the IQR Method



```r
Q1 <- quantile(hotel_data$adr, 0.25)
Q3 <- quantile(hotel_data$adr, 0.75)

IQR_value <- IQR(hotel_data$adr)

lower_limit <- Q1 - 1.5 * IQR_value
upper_limit <- Q3 + 1.5 * IQR_value

lower_limit
upper_limit
```

Count the Outliers

```r
sum(hotel_data$adr < lower_limit | hotel_data$adr > upper_limit)
```

Remove the Outliers

```r
hotel_data <- hotel_data[
  hotel_data$adr >= lower_limit &
  hotel_data$adr <= upper_limit,
]
```

Verifying

```r
boxplot(
  hotel_data$adr,
  main = "ADR After Removing Outliers",
  ylab = "ADR"
)
```

Normalization

```r
hotel_data$adr_normalized <- (hotel_data$adr - min(hotel_data$adr)) /  (max(hotel_data$adr) - min(hotel_data$adr))
```

Verifying

```r
summary(hotel_data$adr_normalized)
```

## **Visualization 2: Booking Cancellation Status**

```r
ggplot(hotel_data, aes(x = factor(is_canceled), fill = factor(is_canceled))) +
  geom_bar() +
  labs(
    title = "Booking Cancellation Status",
    x = "Cancellation Status",
    y = "Number of Bookings",
    fill = "Status"
  ) +
  scale_x_discrete(
    labels = c("Not Cancelled", "Cancelled")
  ) +
  scale_fill_discrete(
    labels = c("Not Cancelled", "Cancelled")
  ) +
  theme_minimal()
```

**Visualization 5: Lead Time vs ADR**

```r
ggplot(hotel_data, aes(x = lead_time, y = adr)) +
  geom_point(alpha = 0.2, size = 0.8) +
  labs(
    title = "Relationship Between Lead Time and Average Daily Rate",
    x = "Lead Time (Days)",
    y = "Average Daily Rate"
  ) +
  theme_minimal()
```

**Visualization 6: Monthly Booking Trend**

```r
monthly_bookings <- hotel_data %>%
  group_by(arrival_date_month) %>%
  summarise(Bookings = n()) %>%
  mutate(
    arrival_date_month = factor(
      arrival_date_month,
      levels = c(
        "January", "February", "March", "April",
        "May", "June", "July", "August",
        "September", "October", "November", "December"
      )
    )
  ) %>%
  arrange(arrival_date_month)
```

```r
ggplot(monthly_bookings,
       aes(x = arrival_date_month,
           y = Bookings,
           group = 1)) +
  geom_line(linewidth = 1) +
  geom_point(size = 3) +
  labs(
    title = "Monthly Booking Trend",
    x = "Arrival Month",
    y = "Number of Bookings"
  ) +
  theme_minimal()
```
